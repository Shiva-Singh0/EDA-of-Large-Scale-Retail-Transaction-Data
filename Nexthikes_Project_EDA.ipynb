# 2. Interquartile Range (IQR) method
### QR tells us the range where the bulk of the values lie. The interquartile range is calculated by subtracting the first quartile from the third quartile. 
### IQR = Q3 - Q1
### Q1 is defined as the middle number between the smallest number and the median of the data set.
### Q2 is the median of the data.
### Q3 is the middle value between the median and the highest value of the data set.

```python
def iqr_outlier_summary(data, columns):
    results = []
    for columns in columns:
        q1 = data[columns].quantile(0.25)
        q3 = data[columns].quantile(0.75)
        iqr = q3 - q1
        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr
        mask = (data[columns] < lower) | (data[columns] > upper)

        results.append({
            "Variable": columns,
            "Q1": q1,
            "Q3": q3,
            "IQR": iqr,
            "Lower Bound": lower,
            "Upper Bound": upper,
            "Outlier Count": int(mask.sum()),
            "Outlier %": mask.mean() * 100
        })
    return pd.DataFrame(results).round(3)

iqr_results = iqr_outlier_summary(EDA, outlier_targets)
display(iqr_results)
```

# Bivariate Analysis

**Bivariate analysis** studies the relationship between two variables.

The project brief asks for:
- Numerical vs numerical
- Numerical vs categorical

For numerical variables, Pearson correlation is used to quantify linear relationships.

## 1. Numerical vs Numerical — Pearson correlation

Correlation ranges from -1 to +1:

- close to +1 → strong positive relationship
- close to -1 → strong negative relationship
- close to 0 → weak/no linear relationship

Correlation does **not** prove causation.

```python
EDA['return_status_encoded'] = EDA['return_status'].map({'Yes': 1, 'No': 0})


# 1. Numerical vs Numerical: Pearson Correlation Matrix

num_cols = ['age', 'product_price', 'quantity', 'discount_percentage', 'final_price', 'delivery_days', 'return_status_encoded']
corr_matrix = EDA[num_cols].corr(method='pearson')

print("=== Pearson Correlation Matrix ===")
display(corr_matrix.round(3))

# Plot Correlation Heatmap
plt.figure(figsize=(14, 12))
sns.heatmap(corr_matrix, annot=True, fmt=".3f", cmap='Blues', cbar=True, square=True)
plt.title('Pearson Correlation Heatmap', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

# 9. Skewness and Kurtosis

### Skewness
Measures asymmetry.

- Positive skew → longer right tail
- Negative skew → longer left tail
- Around 0 → approximately symmetric

### Kurtosis
Describes tail heaviness/extreme observations.

High kurtosis can indicate that extreme values occur more frequently than expected under a normal distribution.

The project brief specifically identifies positive skew in `product_price` and `final_price`, and high kurtosis in `final_price`.

- Direct transaction components are the primary drivers of financial value within the dataset. Product price and order quantity display the strongest relationships with the final transaction value, yielding strong Pearson correlation coefficients of 0.73 and 0.58 respectively.

- **Actionable Insight**: Transaction-value monitoring, revenue forecasting, and customer segmentation strategies should place heavy analytical emphasis on pricing structures and purchasing volume, rather than relying on demographic variables.

### 2. Evaluate Discounts Alongside Transaction Value
- Discount strategies have a measurable, compounding impact on top-line revenue. The data shows that the average final transaction price steadily declines as the discount percentage increases, dropping from approximately ₹76,030 at a 0% discount down to ₹52,051 at a 30% discount.

- **Actionable Insight**: Promotional decisions must be evaluated in direct combination with final transaction values rather than analyzing discount percentages in isolation. Future studies should investigate whether specific discount thresholds yield varying performance outcomes across distinct product categories and customer segments.

### 3. Avoid Using Age as a Primary Transaction-Value Indicator
- Demographic targeting based on age does not correlate with purchasing power in this dataset. Customer age shows virtually zero linear relationship with the final transaction price, resulting in an overall Pearson correlation of -0.0008. This lack of correlation remains consistently flat even when filtering within isolated customer segments.

- **Actionable Insight**: Age should be discarded as a primary indicator or predictor of transaction value. Marketing and sales frameworks should pivot toward behavioral metrics (like past purchase value) rather than age-based demographics.

## ⚙️ Part II: Operational Optimization and Data Integrity Protocols

### 4. Use Multi-Factor Analysis for Return Behavior
- Return rates remain relatively consistent across product categories, payment methods, shipping types, and delivery durations. Because no individual factor showed a strong standalone relationship with returns during bivariate analysis, return behavior cannot be predicted using a single variable.

- **Actionable Insight**: Future investigations should use multi-factor analysis to examine combinations of customer, product, and operational characteristics. Advanced predictive models should also be developed to explore interaction effects and calculate return probabilities.

### 5. Retain Valid High-Value Transactions
- Although the final_price variable contains statistically extreme observations, further investigation reveals that these transactions are entirely consistent with the underlying product price, quantity, and discount structures.

- **Actionable Insight**: These high-value data points should not be removed as outliers. They represent legitimate premium or bulk purchases that are crucial for understanding the true distribution of transaction values.

### 6. Final Analytical Takeaway
- **Value Drivers**: Final transaction value is almost exclusively driven by direct components—specifically product price and order quantity. Conversely, customer demographics (such as age) and operational timelines (such as delivery duration) show no significant standalone relationship with the final price.

- **Discount Dynamics**: Higher discount percentages show a weak negative association with the final transaction value, resulting in a broader decline in average final prices across increasing discount tiers.

- **Return Predictability**: Return behavior is highly stable across individual customer, product, and operational variables. Predicting returns will require composite features and multi-factor modeling rather than relying on a single predictor.

- **Data Integrity**: The analysis underscores the importance of distinguishing statistical outliers from erroneous data. High-value transactions should be preserved because they are internally consistent with the dataset's pricing logic.
